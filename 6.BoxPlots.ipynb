```python
import matplotlib.pyplot as plt
import numpy as np
```

# Box Plot

```python
data = [7, 8, 5, 6, 9, 4, 10, 12, 15]

plt.boxplot(data)
plt.grid(True)
```

### Multiple Datasets

```python

group1 = np.random.normal(50, 10, 100)
group2 = np.random.normal(60, 15, 100)

plt.boxplot([group1, group2], tick_labels = ["group1", "group2"])
plt.grid(True)
plt.show()
```

### Horizontal Box Plot

```python
data = [7, 8, 5, 6, 9, 4, 10, 12, 15]

plt.boxplot(data, vert = False)
plt.grid(True)
```

### Show Mean

```python
data = [7, 8, 5, 6, 9, 4, 10, 12, 15]

plt.boxplot(data, showmeans = True)
plt.grid(True)
```

### Change Whiskers Length

```python
data = [7, 8, 5, 6, 9, 4, 10, 12, 15]

plt.boxplot(data, showmeans = True, whis = 2.0)
plt.grid(True)
```
