```python
import matplotlib.pyplot as plt
```

### Pie Chart

```python
expenses = ["Salaries", "Rent", "Marketing", "R&D", "Miscellaneous"]
amounts = [500, 158, 120, 100, 50]

plt.style.use("dark_background")
plt.pie(amounts, labels = expenses)
plt.style.available
```

```python
expenses = ["Salaries", "Rent", "Marketing", "R&D", "Miscellaneous"]
amounts = [500, 158, 120, 100, 50]

# colors = ["red", "blue", "pink", "black", "yellow"]
# plt.style.use("default")

explode = [0, 0, 0, 0.2, 0]

plt.pie(amounts, labels = expenses, autopct = "%1.1f%%",
         wedgeprops = {
             "edgecolor": "black",
             "linewidth": 1,
             "linestyle": "--",
             "alpha": 1,
             "fill": True
         },
        explode = explode,
        startangle = 360,
        shadow = True
       )
plt.show()
```
