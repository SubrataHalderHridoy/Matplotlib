```python
import matplotlib.pyplot as plt
```

# Bar Charts

```python
movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argu"
]
years = [2008, 2009, 2010, 2011, 2012]
oscar_revenue = [1005, 170, 427, 133, 232]

plt.bar(years, oscar_revenue, color = "purple")
plt.title("Revenue in each year for oscar movies")
plt.xlabel("Year")
plt.ylabel("Revenue (in $M)")

for i in range(len(years)):
   plt.text(years[i], oscar_revenue[i] + 10, str(oscar_revenue[i]), ha = "center")

plt.ylim(0, max(oscar_revenue)+100)
```

```python
import numpy as np

movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argu"
]
oscar_revenue = [1005, 170, 427, 133, 232]

non_oscar_movies = ("Slumodg Milllionaire","Avatar","Inception","Hugo","Lincoln")
non_oscar_revenue = [378, 2788, 829, 185, 275]

years = [2008, 2009, 2010, 2011, 2012]

x=np.arange(len(years))
width = 0.4

plt.bar(x-width/2, oscar_revenue, width, label = "Oscar Movies") # shifting to left
plt.bar(x+width/2, non_oscar_revenue, width, label = "Non-Osacr Movies") # shifting to right

plt.title("Oscar vs Non-Oscar Movies Revenue")
plt.xlabel("Year")
plt.ylabel("Revenue($M)")
plt.legend()

plt.xticks(x, years)
```
